## A.4 Automatic differentiation made easy

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/8.webp" width="600px">

```python
import torch.nn.functional as F     # นำเข้าฟังก์ชันสำเร็จรูปสำหรับ Deep Learning (เช่น Loss Function ต่าง ๆ)
from torch.autograd import grad     # นำเข้าฟังก์ชัน grad เพื่อใช้คำนวณหาค่า Gradient (อนุพันธ์/ดริฟ) แบบเฉพาะเจาะจง

# --- 1. กำหนดข้อมูลนำเข้าและตัวแปรในระบบ (Forward Input Setup) ---
y = torch.tensor([1.0])     # คำตอบที่ถูกต้องจริง ๆ (Label / Target) ของข้อมูลชุดนี้
x1 = torch.tensor([1.1])    # ข้อมูลดิบที่ป้อนเข้าสู่โมเดล (Input Feature)

# กำหนด Weight (w1) และ Bias (b) พร้อมเปิดระบบ requires_grad=True เพื่อให้ PyTorch คอยบันทึกประวัติการคำนวณสำหรับทำ Backpropagation
w1 = torch.tensor([2.2], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)

# --- 2. ขั้นตอนการคำนวณขาไปเพื่อหาผลลัพธ์และความผิดพลาด (Forward Pass & Loss) ---
z = x1 * w1 + b         # คำนวณสมการเส้นตรง (Net Input): (1.1 * 2.2) + 0.0 = 2.42
a = torch.sigmoid(z)    # ส่งเข้าฟังก์ชัน Sigmoid เพื่อแปลงค่าให้อยู่ในช่วงความน่าจะเป็น (0 ถึง 1)

# คำนวณค่าความผิดพลาดของโมเดลด้วย Binary Cross Entropy Loss (เทียบผลลัพธ์ที่โมเดลเดา 'a' กับคำตอบจริง 'y')
loss = F.binary_cross_entropy(a, y)

# --- 3. ขั้นตอนคำนวณขากลับเพื่อหาความชัน (Backward Pass / Gradient Calculation) ---
# หาค่าดริฟย้อนกลับของ Loss เทียบกับน้ำหนัก w1 (∂Loss / ∂w1)
# ใส่ retain_graph=True เพื่อล็อกกราฟการคำนวณเอาไว้ ไม่ให้ระบบทำลายทิ้งหลังจากบรรทัดนี้ทำงานเสร็จ
grad_L_w1 = grad(loss, w1, retain_graph=True)

# หาค่าดริฟย้อนกลับของ Loss เทียบกับค่า Bias b (∂Loss / ∂b) โดยใช้กราฟการคำนวณที่ล็อกค้างไว้จากบรรทัดก่อนหน้า
grad_L_b = grad(loss, b, retain_graph=True)

# --- 4. แสดงผลลัพธ์ค่า Gradient ที่คำนวณได้ ---
print(grad_L_w1)        # พิมพ์ค่าความชันเพื่อดูทิศทางและขนาดที่ต้องใช้อัปเดตค่าน้ำหนัก w1
print(grad_L_b)         # พิมพ์ค่าความชันเพื่อดูทิศทางและขนาดที่ต้องใช้อัปเดตค่า Bias b
```

```python
# --- 3. ขั้นตอนคำนวณขากลับอัตโนมัติด้วย Backward (Alternative to torch.autograd.grad) ---
# สั่งให้ PyTorch คำนวณหาค่า Gradient (ดริฟย้อนกลับ) ของ Loss เทียบกับตัวแปรทุกตัวที่เปิด requires_grad=True พร้อมกันในคำสั่งเดียว
# ผลลัพธ์ที่ได้จากการคำนวณจะถูกนำไปอัปเดตและ "เก็บซ่อนไว้ภายใน" คุณสมบัติ .grad ของตัวแปร w1 และ b โดยตรง
loss.backward()

# --- 4. แสดงผลลัพธ์ค่า Gradient ที่ถูกบันทึกไว้ในตัวแปร ---
print(w1.grad)   # ดึงค่าความชัน (∂Loss / ∂w1) ที่คำนวณได้จากคำสั่ง backward() ออกมาพิมพ์ดูตัวเลข
print(b.grad)    # ดึงค่าความชัน (∂Loss / ∂b) ที่คำนวณได้จากคำสั่ง backward() ออกมาพิมพ์ดูตัวเลข
```

## A.5 Implementing multilayer neural networks

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/9.webp" width="500px">

```python
'''class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(

            # 1st hidden layer
            torch.nn.Linear(num_inputs, 30),
            torch.nn.ReLU(),

            # 2nd hidden layer
            torch.nn.Linear(30, 20),
            torch.nn.ReLU(),

            # output layer
            torch.nn.Linear(20, num_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits'''
```

```python
# สร้างคลาสโมเดลโครงข่ายประสาทเทียม โดยสืบทอดคุณสมบัติมาจาก torch.nn.Module ซึ่งเป็นโครงสร้างมาตรฐานของ PyTorch
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        # เรียกใช้งานฟังก์ชัน __init__ ของคลาสแม่ (torch.nn.Module) เพื่อตั้งค่าระบบภายในของ PyTorch
        super().__init__()

        # torch.nn.Sequential ใช้สำหรับมัดรวมเลเยอร์ต่าง ๆ เข้าด้วยกัน เพื่อให้ข้อมูลไหลผ่านตามลำดับจากบนลงล่าง
        self.layers = torch.nn.Sequential(

            # --- เลเยอร์ซ่อนชั้นที่ 1 (1st Hidden Layer) ---
            # รับข้อมูลขาเข้าจำนวน num_inputs มิติ แล้วขยายออกเป็น 30 นิวรอน (Node)
            torch.nn.Linear(num_inputs, 30),
            # เปิดใช้งานฟังก์ชันกระตุ้น ReLU เพื่อตัดค่าที่ติดลบให้กลายเป็น 0 ช่วยให้โมเดลสามารถเรียนรู้ความสัมพันธ์ที่ซับซ้อน (Non-linear) ได้
            torch.nn.ReLU(),

            # --- เลเยอร์ซ่อนชั้นที่ 2 (2nd Hidden Layer) ---
            # รับข้อมูลต่อมาจากชั้นแรกจำนวน 30 มิติ แล้วบีบ/ปรับให้เหลือ 20 นิวรอน
            torch.nn.Linear(30, 20),
            # ผ่านฟังก์ชันกระตุ้น ReLU อีกครั้งเพื่อคัดกรองข้อมูลก่อนส่งต่อ
            torch.nn.ReLU(),

            # --- เลเยอร์ผลลัพธ์ (Output Layer) ---
            # รับข้อมูล 20 มิติสุดท้ายเข้ามา แล้วแปลงให้เหลือจำนวนมิติเท่ากับคำตอบที่เราต้องการ (num_outputs)
            torch.nn.Linear(20, num_outputs),
        )

    # ฟังก์ชัน forward กำหนดเส้นทางการไหลของข้อมูลเมื่อเราป้อนข้อมูล (Input 'x') เข้าสู่โมเดล
    def forward(self, x):
        # ส่งข้อมูล x วิ่งทะลุผ่านเลเยอร์ต่าง ๆ ที่เรามัดรวมไว้ใน self.layers ตามลำดับ
        # ผลลัพธ์ที่ได้เรียกว่า 'logits' ซึ่งคือค่าตัวเลขดิบที่ยังไม่ได้ถูกแปลงเป็นความน่าจะเป็น
        logits = self.layers(x)

        # ส่งค่าคะแนนดิบ (logits) ออกไป เพื่อนำไปคำนวณ Loss หรือแปลงเป็นความน่าจะเป็นด้วย Softmax ต่อไป
        return logits
```

```python
# --- การสร้างโมเดลตัวจริงเพื่อใช้งาน (Model Instantiation) ---
# ทำการสร้างออบเจกต์โมเดล (Model Instance) ขึ้นมาจากคลาส NeuralNetwork ที่เราเขียนโครงสร้างไว้
# โดยกำหนดสเปกของโครงข่ายประสาทเทียมตัวนี้ดังนี้:
#   - 50 (num_inputs): โมเดลนี้จะรองรับข้อมูลดิบขาเข้าที่มีทั้งหมด 50 มิติ (หรือมี 50 คุณลักษณะ/Features)
#   - 3 (num_outputs): โมเดลนี้จะคำนวณและพ่นผลลัพธ์ออกมาเป็นค่าคะแนน (Logits) จำนวน 3 ตัวเลข
#                      (เช่น เหมาะสำหรับงานคัดแยกประเภทข้อมูลออกเป็น 3 หมวดหมู่/Classes)
model = NeuralNetwork(50, 3)
```

```python
print(model)
```

```python
# --- การนับจำนวนพารามิเตอร์ทั้งหมดในโมเดล (Counting Trainable Parameters) ---
# ใช้ List Comprehension ร่วมกับฟังก์ชัน sum() เพื่อคำนวณหาจำนวนตัวเลขค่าน้ำหนัก (Weights และ Biases) ทั้งหมดในโมเดล
#   - model.parameters(): ดึงตัวแปรค่าน้ำหนักทั้งหมดในทุก ๆ เลเยอร์ของโมเดลออกมาวนลูป
#   - if p.requires_grad: เลือกนับเฉพาะตัวแปรที่เปิดระบบให้ฝึกฝนได้ (Trainable) เท่านั้น (ตัวไหนที่ถูกล็อกค่าไว้จะไม่ถูกนำมารวม)
#   - p.numel(): ย่อมาจาก "Number of Elements" ทำหน้าที่นับจำนวนเม็ดตัวเลขที่อยู่ภายในเทนเซอร์ของเลเยอร์นั้น ๆ
num_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

# พิมพ์จำนวนตัวแปรทั้งหมดที่โมเดลนี้ต้องใช้ในการเรียนรู้และปรับค่าในตอนฝึกฝน (Training)
print("Total number of trainable model parameters:", num_params)
```

```python
# --- 1. ล็อกผลการสุ่มตัวเลข (Setting Random Seed for Reproducibility) ---
# กำหนดค่า Seed เป็น 123 เพื่อให้การสุ่มตัวเลขค่าน้ำหนัก (Weights) ของ PyTorch ในคำสั่งถัดไปได้ผลลัพธ์เหมือนเดิมทุกครั้งที่รันโค้ด
# ประโยชน์คือช่วยให้เราสามารถทดลอง ตรวจสอบ และเปรียบเทียบผลลัพธ์ของโค้ดกับคนอื่น (หรือโค้ดในหนังสือ) ได้อย่างแม่นยำ
torch.manual_seed(123)

# --- 2. สร้างโมเดลตัวใหม่ (Model Instantiation) ---
# สร้างโมเดลใหม่ขึ้นมา ซึ่งในขั้นตอนนี้เลเยอร์ต่าง ๆ จะถูกสุ่มค่าเริ่มต้น (Initial Weights) ขึ้นมาตามโครงสร้างที่ล็อกด้วย Seed ด้านบน
model = NeuralNetwork(50, 3)

# --- 3. ตรวจดูค่าน้ำหนักในเลเยอร์แรกสุด (Inspecting First Layer Weights) ---
#   - model.layers[0]: เจาะจงเลือกเข้าถึงเลเยอร์แรกสุดที่อยู่ข้างในตระกูล Sequential (นั่นคือชั้น torch.nn.Linear(50, 30))
#   - .weight: ดึงเมทริกซ์ค่าน้ำหนัก (Weight Matrix) ขนาด 30 แถว x 50 คอลัมน์ ออกมาเพื่อพิมพ์ดูค่าตัวเลขข้างใน
print(model.layers[0].weight)
```

```python
# --- 1. ล็อกผลการสุ่มตัวเลข (Setting Random Seed) ---
# กำหนดค่า Seed เป็น 123 เพื่อให้คำสั่งสุ่มตัวเลขในบรรทัดถัดไป (torch.rand) พ่นตัวเลขชุดเดิมออกมาทุกครั้งที่รันโค้ด
torch.manual_seed(123)

# --- 2. สร้างข้อมูลทดสอบจำลอง (Creating Dummy Input Data) ---
#   - torch.rand((1, 50)): สุ่มสร้างตัวเลขทศนิยมที่มีค่าระหว่าง 0 ถึง 1 ขึ้นมาเป็น 2D Tensor ขนาด 1 แถว x 50 คอลัมน์
#   - เปรียบเสมือนการจำลองข้อมูลตัวอย่างขึ้นมา 1 ข้อมูล (Batch Size = 1) โดยที่ข้อมูลนั้นมีจำนวนคุณลักษณะหรือตัวแปรทั้งหมด 50 มิติ
X = torch.rand((1, 50))

# --- 3. ป้อนข้อมูลเข้าโมเดลเพื่อคำนวณผลลัพธ์ (Forward Pass) ---
# ส่งเทนเซอร์ข้อมูล X วิ่งทะลุผ่านโครงข่ายประสาทเทียมที่เราสร้างไว้ (model) เพื่อให้เลเยอร์ต่าง ๆ คำนวณคณิตศาสตร์ออกมาเป็นผลลัพธ์
# ผลลัพธ์ที่ได้จะถูกเก็บไว้ในตัวแปร 'out' ซึ่งจะอยู่ในรูปของค่าคะแนนดิบ (Logits) จำนวน 3 ตัวเลข (ตามโครงสร้างเอาท์พุตที่เรากำหนดไว้)
out = model(X)

# --- 4. แสดงผลลัพธ์คะแนนทำนาย ---
# พิมพ์ค่าคะแนนดิบ 3 ตัวที่โมเดลนี้คำนวณได้ออกมาดูบนหน้าจอ
print(out)
```

```python
# --- การรันโมเดลในโหมดปิดระบบคำนวณความชัน (Inference Mode with no_grad) ---
# ใช้คำสั่ง 'with torch.no_grad():' เพื่อบอกให้ PyTorch ปิดระบบการสร้าง Computation Graph และหยุดบันทึกประวัติการคำนวณทั้งหมดชั่วคราว
#   - ประโยชน์ที่ 1: ช่วยประหยัดหน่วยความจำ (RAM/VRAM) ได้มหาศาล เพราะไม่มีการเก็บค่าเพื่อใช้ทำ Backpropagation
#   - ประโยชน์ที่ 2: ทำให้โมเดลประมวลผลคำนวณผลลัพธ์ (Forward Pass) ได้รวดเร็วขึ้นอย่างมาก
#   - เหมาะสำหรับ: ขั้นตอนการนำโมเดลไปทดสอบ (Evaluation), ตรวจวัดประสิทธิภาพ (Validation) หรือนำไปใช้งานจริง (Inference)
with torch.no_grad():

    # ป้อนข้อมูล X เข้าโมเดลเพื่อคำนวณผลลัพธ์ในโหมดประหยัดพลังงาน
    out = model(X)

# พิมพ์ค่าคะแนนดิบ (Logits) 3 ตัวที่ได้จากการคำนวณออกมาดูบนหน้าจอ ซึ่งจะได้ตัวเลขเท่ากับการรันปกติทุกประการ
print(out)
```

```python
# --- การแปลงผลลัพธ์โมเดลเป็นเปอร์เซ็นต์ความน่าจะเป็น (Converting Logits to Probabilities) ---
# ใช้โหมดประหยัดหน่วยความจำ (no_grad) เพื่อคำนวณผลลัพธ์อย่างรวดเร็วสำหรับการนำไปใช้งานจริง (Inference)
with torch.no_grad():

    # 1. model(X) จะคำนวณและพ่นค่าคะแนนดิบออกมา 3 ตัว (Logits) เช่น [2.1, -0.5, 0.8]
    # 2. torch.softmax(..., dim=1) จะนำคะแนนดิบเหล่านั้นมาเข้าสูตรแปลงให้กลายเป็นค่าความน่าจะเป็น (Probability)
    #    - ผลลัพธ์ที่ได้จะเป็นตัวเลขทศนิยม 3 ตัวที่อยู่ระหว่าง 0 ถึง 1 เสมอ
    #    - ผลรวมของตัวเลขทั้ง 3 ตัวนี้เมื่อนำมาบวกกันจะได้เท่ากับ 1.0 (หรือ 100%) พอดีเป๊ะ
    #    - dim=1 หมายถึง ให้คำนวณเปรียบเทียบหาค่า Softmax แนวนอนตามแนวคอลัมน์ (คำนวณแยกในแต่ละข้อมูลขาเข้า)
    out = torch.softmax(model(X), dim=1)

# พิมพ์เปอร์เซ็นต์ความน่าจะเป็นของทั้ง 3 หมวดหมู่ที่โมเดลนี้ทำนายได้ออกมาดูบนหน้าจอ
print(out)
```

```python
from torch.utils.data import Dataset  # นำเข้าคลาสแม่ (Base Class) สำหรับการจัดการชุดข้อมูลของ PyTorch


# สร้างคลาส ToyDataset โดยสืบทอดคุณสมบัติ (Inherit) มาจากคลาส Dataset ของ PyTorch
class ToyDataset(Dataset):

    # 1. ฟังก์ชันตัวสร้าง (Constructor): ใช้สำหรับรับข้อมูลเข้ามาเก็บไว้ในคลาสเมื่อทำการสร้างออบเจกต์
    def __init__(self, X, y):
        self.features = X             # เก็บข้อมูลคุณลักษณะขาเข้า (เช่น X_train หรือ X_test) ไว้ในตัวแปรประจำตัว
        self.labels = y               # เก็บข้อมูลคำตอบเฉลย (เช่น y_train หรือ y_test) ไว้ในตัวแปรประจำตัว

    # 2. ฟังก์ชันดึงข้อมูลตามตำแหน่ง (Indexing): ถูกเรียกใช้อัตโนมัติเมื่อเราพิมพ์คำสั่งดึงข้อมูลด้วยวงเล็บเหลี่ยม เช่น dataset[index]
    def __getitem__(self, index):
        one_x = self.features[index]  # ดึงข้อมูลขาเข้าชิ้นที่ระบุตามค่า index ออกมา 1 ตัวอย่าง
        one_y = self.labels[index]    # ดึงเฉลยที่คู่กับข้อมูลชิ้นนั้นออกมา 1 ตัวอย่าง
        return one_x, one_y           # ส่งข้อมูลตัวอย่างและเฉลยคู่กันกลับออกไปเป็นคู่ (Tuple)

    # 3. ฟังก์ชันนับจำนวนข้อมูลทั้งหมด (Length): ถูกเรียกใช้อัตโนมัติเมื่อเราสั่งรันคำสั่ง len(dataset)
    def __len__(self):
        # นับจำนวนแถวทั้งหมดของเฉลย (ขนาดมิติแรกสุด .shape[0]) เพื่อรายงานว่าชุดข้อมูลนี้มีข้อมูลอยู่ทั้งหมดกี่ตัวอย่าง
        return self.labels.shape[0]

# --- การใช้งานจริง (Instantiation) ---
# ทำการนำข้อมูลดิบ (X_train, y_train และ X_test, y_test) มาบรรจุลงในคลาส ToyDataset
# เพื่อเปลี่ยนให้ข้อมูลดิบธรรมดากลายเป็น "วัตถุชุดข้อมูลของ PyTorch" ที่พร้อมนำไปใช้งานต่อร่วมกับ DataLoader
train_ds = ToyDataset(X_train, y_train)  # ชุดข้อมูลสำหรับฝีกฝน
test_ds = ToyDataset(X_test, y_test)    # ชุดข้อมูลสำหรับทดสอบ
```

```python
from torch.utils.data import DataLoader  # นำเข้าคลาส DataLoader ซึ่งเป็นระบบจัดการและลำเลียงข้อมูลของ PyTorch

# ล็อกผลการสุ่มตัวเลข (Seed) เป็น 123 เพื่อให้ขั้นตอนการสับไพ่สลับลำดับข้อมูล (shuffle) ในคำสั่งถัดไป
# พ่นลำดับข้อมูลชุดเดิมออกมาทุกครั้งที่รันโค้ด ช่วยให้ผลลัพธ์คงที่ตรวจสอบง่าย (Reproducibility)
torch.manual_seed(123)

# --- การสร้างระบบลำเลียงข้อมูลสำหรับฝึกฝน (Setting up Train DataLoader) ---
# ทำการห่อหุ้มวัตถุชุดข้อมูล train_ds ไว้ด้วย DataLoader เพื่อให้ได้ความสามารถในการจัดระเบียบข้อมูลเพิ่มขึ้น ดังนี้:
train_loader = DataLoader(
    dataset=train_ds,      # กำหนดวัตถุชุดข้อมูลต้นทางที่ต้องการให้ดึงข้อมูลออกมา (ในที่นี้คือชุดข้อมูลเทรนที่เราสร้างไว้)
    batch_size=2,         # กำหนดให้จัดแบ่งข้อมูลส่งออกไปประมวลผลเป็นชุดย่อย ๆ ครั้งละ 2 ตัวอย่าง (Batch Size = 2)
    shuffle=True,         # สั่งให้ "สลับลำดับข้อมูลทุกครั้ง" ที่เริ่มรอบการเทรนใหม่ (Epoch) เพื่อป้องกันไม่ให้โมเดลจดจำลำดับก่อนหลังของข้อสอบ
    num_workers=0         # กำหนดจำนวนคอร์ประมวลผล (CPU Threads) ที่ใช้ในการดึงข้อมูลขนานกัน
                          # เลข 0 หมายถึงให้ทำงานบน Thread หลักตัวเดียว (เหมาะสำหรับโค้ดเริ่มต้น หรือรันบน Windows เพื่อป้องกันปัญหาระบบค้าง)
)
```

```python
import torch.nn.functional as F  # นำเข้าฟังก์ชันสำเร็จรูป (เช่น Cross Entropy Loss)

# --- 1. ตั้งค่าเริ่มต้นและสร้างระบบ Optimizer ---
torch.manual_seed(123)  # ล็อกผลการสุ่มเริ่มต้นเพื่อให้ผลการรันคงที่ทุกครั้ง
model = NeuralNetwork(num_inputs=2, num_outputs=2)  # สร้างโมเดลตัวจริงที่รับอินพุต 2 มิติ และพ่นเอาท์พุต 2 มิติ

# สร้างตัวปรับค่าน้ำหนัก (Optimizer) ชนิด SGD (Stochastic Gradient Descent)
#   - model.parameters(): ส่งตัวแปร Weights และ Biases ทั้งหมดในโมเดลไปให้ดูแล
#   - lr=0.5 (Learning Rate): อัตราการเรียนรู้เท่ากับ 0.5 (กำหนดความใหญ่ของก้าวในการปรับค่าน้ำหนัก)
optimizer = torch.optim.SGD(model.parameters(), lr=0.5)

num_epochs = 3  # กำหนดจำนวนรอบใหญ่ในการเรียนรู้ (รันจนครบข้อมูลทั้งหมด 3 รอบ)

# --- 2. เริ่มต้นลูปการฝึกฝน (Training Loop) ---
for epoch in range(num_epochs):

    model.train()  # สั่งเปิดโหมดฝึกฝน (Training Mode) เพื่อเปิดใช้งานระบบคำนวณและเตรียมปรับค่าน้ำหนักภายในโมเดล

    # วนลูปดึงข้อมูลย่อย (Batches) ออกมาจาก train_loader ทีละชุดจนครบทั้งเซต
    for batch_idx, (features, labels) in enumerate(train_loader):

        # ขั้นตอนที่ 1: Forward Pass - ป้อนข้อมูลฟีเจอร์เข้าโมเดลเพื่อคำนวณหาคะแนนดิบ (Logits)
        logits = model(features)

        # ขั้นตอนที่ 2: Compute Loss - คำนวณหาค่าความผิดพลาดด้วย Cross Entropy
        # (เทียบค่าคะแนนดิบจากโมเดล 'logits' กับคำตอบเฉลยจริง 'labels')
        loss = F.cross_entropy(logits, labels)

        # ขั้นตอนที่ 3: Zero Gradients - สั่งล้างค่าความชันเก่าที่ค้างอยู่ในตัวแปรออกให้กลายเป็น 0
        # (หากไม่ล้าง PyTorch จะเอาความชันรอบเก่ามาบวกทบกับรอบใหม่ ทำให้โมเดลเรียนรู้ผิดทิศทาง)
        optimizer.zero_grad()

        # ขั้นตอนที่ 4: Backward Pass - สั่งดิฟย้อนกลับเพื่อหาค่าความชัน (Gradients) ของพารามิเตอร์ทุกตัวในโมเดล
        loss.backward()

        # ขั้นตอนที่ 5: Update Weights - สั่งให้ Optimizer นำค่าความชันที่ได้ไปปรับจูนค่าน้ำหนักจริงในตัวแปรทันที
        optimizer.step()

        ### --- 3. บันทึกและแสดงผลระหว่างเทรน (Logging) ---
        # พิมพ์ความคืบหน้า: รอบเอพ็อกปัจจุบัน, หมายเลขบัคช์, และค่าความผิดพลาด (Loss) ที่ลดลงเรื่อย ๆ
        print(f"Epoch: {epoch+1:03d}/{num_epochs:03d}"
              f" | Batch {batch_idx+1:03d}/{len(train_loader):03d}"
              f" | Train/Val Loss: {loss:.2f}")

    # --- 4. สลับเข้าสู่โหมดประเมินผลท้ายรอบ ---
    model.eval()  # สั่งเปิดโหมดทดสอบ (Evaluation Mode) เพื่อเตรียมนำโมเดลไปทดสอบกับข้อสอบจริง (test_loader)
    # (โค้ดสำหรับวัดผลการเรียนรู้ของรอบนี้จะเขียนต่อท้ายตรงนี้)
```

```python
# --- สร้างฟังก์ชันสำหรับคำนวณความแม่นยำของโมเดล (Defining Accuracy Function) ---
def compute_accuracy(model, dataloader):

    # 1. ตั้งค่าโมเดลให้อยู่ในโหมดประเมินผล (Evaluation Mode)
    # เพื่อสั่งปิดระบบบางตัวที่ใช้เฉพาะตอนเทรน เช่น Dropout และดึงพารามิเตอร์มาใช้วัดผลอย่างเสถียร
    model.eval()

    correct = 0.0          # ตัวแปรสำหรับเก็บสะสมจำนวนข้อที่โมเดลตอบถูกทั้งหมด (เริ่มต้นที่ 0)
    total_examples = 0     # ตัวแปรสำหรับนับจำนวนข้อมูลทั้งหมดที่ส่งเข้ามาทดสอบ (เริ่มต้นที่ 0)

    # 2. วนลูปดึงข้อมูลย่อย (Batches) ออกมาจาก DataLoader ที่เราส่งเข้ามาในฟังก์ชันทีละชุดจนครบ
    for idx, (features, labels) in enumerate(dataloader):

        # เปิดบล็อกประหยัดหน่วยความจำ (no_grad) เพื่อคำนวณหาผลลัพธ์โดยไม่ต้องบันทึกค่าความชัน (Gradients)
        with torch.no_grad():
            logits = model(features) # ป้อนฟีเจอร์เข้าโมเดลเพื่อคำนวณหาค่าคะแนนดิบ (Logits)

        # ค้นหาตำแหน่งคอลัมน์ที่มีคะแนนสูงสุดในแต่ละแถว เพื่อสรุปเป็นหมวดหมู่ที่โมเดลเลือกทำนาย [10.10]
        predictions = torch.argmax(logits, dim=1)

        # เปรียบเทียบผลการทำนาย (predictions) กับเฉลยจริง (labels) แบบตัวต่อตัว จะได้เทนเซอร์ค่า True/False
        compare = labels == predictions

        # นับจำนวนค่า True (ที่ตอบถูก) ใน Batch นี้ โดยแปลง True เป็น 1 แล้วนำไปบวกสะสมรวมไว้ในตัวแปร correct
        correct += torch.sum(compare)

        # นับจำนวนข้อมูลทั้งหมดใน Batch นี้ (len ของเทนเซอร์เปรียบเทียบ) แล้วนำไปบวกสะสมไว้ในตัวแปร total_examples
        total_examples += len(compare)

    # 3. คำนวณหาค่าสัดส่วนความแม่นยำโดยนำ (จำนวนข้อที่ถูกทั้งหมด / จำนวนข้อมูลทั้งหมด)
    # และใช้คำสั่ง .item() เพื่อแปลงมูลค่าจากวัตถุ Tensor ของ PyTorch ออกมาเป็นตัวเลขทศนิยมปกติของ Python
    return (correct / total_examples).item()
```
